```
from ultralytics import YOLO

# Ultralytics automatically downloads SKU-110K (11,743 images, 1.73M bounding boxes)
model = YOLO('yolov8m.pt')
results = model.train(
    data='SKU-110K.yaml',
    epochs=35,
    imgsz=1024,
    batch=16,
    device=0,
    amp=True,
    cos_lr=True,
    max_det=600,
    iou=0.55,
    project='plangograph_sku110k',
    name='yolov8m_dense_1024'
)
model.export(format='onnx', dynamic=True, simplify=True)
```
